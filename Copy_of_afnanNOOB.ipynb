```python

from chatterbot import ChatBot
from chatterbot.trainers import ListTrainer
import logging

logging.basicConfig(level=logging.WARNING)


bot = ChatBot(
    "Nana",
    storage_adapter='chatterbot.storage.SQLStorageAdapter',
    logic_adapters=[
        'chatterbot.logic.TimeLogicAdapter',
        'chatterbot.logic.MathematicalEvaluation',
        'chatterbot.logic.BestMatch'
    ],

    read_only=False
)

trainer = ListTrainer(bot)


trainer.train(
    [

     'Hi',
     'Hello',
     'Hello',
     'Hi there!',


     'How are you?',
     'I am fine, thank you. What about you?',
     'I am also fine.',
     'Nice to hear that.',



     'What is your name?',
     'My name is Nana. I am a super bocah chatbot.',
     'Who are you?',
     'I am Nana, a very pro ml player.',
     'Tell me about yourself.',
     'I am a virtual assistant created to chat with you.',


     'Thank you',
     'You are most welcome!',


     'Apa khabar?',
     'Saya baik-baik sahaja, terima kasih.',
     'Selamat pagi',
     'Selamat pagi juga.',
     'dah mandi ke',
     'Maaf, saya tidak boleh mandi. Saya adalah program komputer!',
     'terima kasih',
     'Sama-sama.',
    ]
)

print("Training complete!")


name = input("Enter your name: ")

print("\nChat started. Type 'bye' to exit.")

while True:
  try:
    print(name, end="-->")
    user_input = input().strip()

    if user_input.lower() == 'bye':
      print("BOT--> Goodbye! Have a great day.")
      break


    bot_response = bot.get_response(user_input)

    print("BOT-->", bot_response)
    print("")

  except (KeyboardInterrupt, EOFError, SystemExit):
    print("\nChat ended.")
    break
```
